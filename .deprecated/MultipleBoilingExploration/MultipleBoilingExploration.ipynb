```python
# KDE 
print(x)
# Compute inter-peak intervals
inter_peak_intervals = np.diff(x)

print(inter_peak_intervals)

# KDE on intervals
kde = gaussian_kde(inter_peak_intervals)
x_grid = np.linspace(0, np.max(inter_peak_intervals), 1000)

plt.plot(x_grid, kde(x_grid))
plt.title("KDE of Inter-Peak Intervals")
plt.xlabel("Interval (s)")
plt.ylabel("Density")
plt.show()
```
