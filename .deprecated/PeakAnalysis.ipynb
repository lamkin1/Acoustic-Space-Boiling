```python
import numpy as np
import pandas as pd
import scipy.signal as signal
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
def clamp(value, lower=0.015, upper=0.1):
    return max(lower, min(value, upper))


def get_peaks(acceleration0):
    percentile = np.percentile(acceleration0, 99.5)
    percent_of_max = 0.1*np.max(acceleration0)

    height = clamp(max(percentile, percent_of_max))
    distance = 350 + 5 / height
    # print(round(height, 4),", ", round(distance))
    # print({0: "percentile", 1: "percent of max", 2: "min threshold"}[np.argmax([percentile, percent_of_max, 0.01])])

    x, y = signal.find_peaks(acceleration0, distance=distance, height=height)
    x = x / 10000
    y = y["peak_heights"]  # unpack data
    return x, y


def plot_file(file):
    data = pd.read_csv(file, index_col="Time")
    time = data.index
    acceleration0 = data['0']

    x, y = get_peaks(acceleration0)

    plt.figure(figsize=(10, 6))
    plt.plot(time, acceleration0, label='Accelerator 0', color='blue', alpha=0.7)
    #plt.plot(time, acceleration1, label='Accelerator 1', color='orange', alpha=0.7)
    plt.plot(x, y, "x", color="orange")
    plt.title(f'Time Domain Signals for {file}')
    plt.xlabel('Time (s)')
    plt.ylabel('Acceleration')
    plt.legend()
    plt.grid(True)
    # plt.axhline(y=height, color='orange', linestyle=':')
    # plt.axhline(y=percentile, color='red', linestyle='--')
    # plt.axhline(y=percent_of_max, color='yellow', linestyle='--')
    # plt.axhline(y=0.01, color='yellow', linestyle='--')
    plt.show()
    return



# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 2-25 PM Thu, Oct 24, 2024 Run6 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 2-40 PM Fri, Jun 14, 2024 Run15 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 2-15 PM Thu, Oct 31, 2024 Run8 .csv')


# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 1-15 PM Thu, Aug 22, 2024 Run5 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 12-59 PM Thu, Aug 22, 2024 Run0.csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 3-44 PM Thu, Aug 29, 2024 Run1 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 1-36 PM Tue, Sep 10, 2024 Run18 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 3-52 PM Thu, Aug 29, 2024 Run2 .csv')

# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 2-52 PM Tue, Oct 15, 2024 Run4 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 3-38 PM Thu, Oct 17, 2024 Run8 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 1-36 PM Thu, Nov 7, 2024 Run14 .csv')
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 4-29 PM Thu, Sep 5, 2024 Run9 .csv')

# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 2-48 PM Fri, Jun 14, 2024 Run16 .csv') ###
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 1-29 PM Thu, Nov 7, 2024 Run12 .csv') ### horizontal
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 4-29 PM Thu, Oct 10, 2024 Run8 .csv') ###
# plot_file('/Users/zach/Desktop/Acoustic-Space-Boiling/Data/After_May/MATLAB 4-09 PM Mon, Aug 26, 2024 Run7 .csv') ###
```
